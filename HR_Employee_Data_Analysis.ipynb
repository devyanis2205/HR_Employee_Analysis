##Step 3: Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

##Q1.How many employees are there in each department?

```python
dept_count = employees["Department"].value_counts()

plt.figure(figsize=(10, 5))
dept_count.plot(kind="bar")

plt.title("Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##Analysis:
 This chart shows the number of employees working in each department. It helps identify which departments have more employees and which have fewer.

##Conclusion:
 The chart helps HR understand employee distribution and plan workforce requirements across departments.

##Q2..How many male and female employees are there?

```python
gender_count = employees["Gender"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution of Employees")
plt.show()
```

##Analysis:
 This chart displays the percentage of employees in each gender category.

##Conclusion:
 The chart helps HR understand the gender composition of the workforce.

##Q3.What is the distribution of employee net salaries, and are there any salary outliers?

```python
plt.figure(figsize=(10, 5))

sns.boxplot(data=salary, x="Net_Salary")

plt.title("Distribution of Employee Net Salaries")
plt.xlabel("Net Salary")
plt.tight_layout()
plt.show()
```

##Analysis:
 The box plot shows the distribution of employee net salaries, including the median, spread of salaries, and potential outliers. The points outside the whiskers represent potential salary outliers, indicating employees with unusually high or low salaries compared to others.
##Conclusion:
 The box plot helps identify salary variations and potential outliers in the dataset. This analysis can help the HR department review unusual salary records and understand overall salary distribution.

##Q4.How many employees work in each city?

```python
city_count = employees["City"].value_counts()

plt.figure(figsize=(10, 5))
city_count.plot(kind="bar")

plt.title("Number of Employees in Each City")
plt.xlabel("City")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##Analysis:
  This chart shows the number of employees located in different cities.

##Conclusion:
  The chart helps HR understand employee distribution across locations.

##Q5.What is the monthly salary expenditure?

```python
monthly_salary = salary.groupby("Salary_Month")["Net_Salary"].sum()
monthly_salary = monthly_salary.sort_index()

plt.figure(figsize=(10, 5))
plt.plot(monthly_salary.index, monthly_salary.values, marker="o")

plt.title("Monthly Salary Expenditure")
plt.xlabel("Month")
plt.ylabel("Total Net Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##Analysis:
 The line chart shows how total net salary expenditure changes from month to month.

##Conclusion:
This chart helps HR monitor salary expenditure and plan monthly budgets.

##Q6.How many attendance records are marked present, absent, or on leave?

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=attendance,
    x="Attendance_Status",
    order=attendance["Attendance_Status"].value_counts().index
)

plt.title("Attendance Status Distribution")
plt.xlabel("Attendance Status")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

##Analysis:
 This chart shows the number of attendance records for each status, such as present, absent, or leave.

##Conclusion:
 The chart helps HR monitor attendance patterns and identify attendance-related issues.

##Q7.What correlations exist between basic salary, allowances, bonuses, deductions, and net salary?

```python
# Select numerical salary columns
salary_columns = [
    "Basic_Salary",
    "Allowances",
    "Bonus",
    "Deductions",
    "Net_Salary"
]

# Calculate correlation matrix
correlation = salary[salary_columns].corr()

# Create heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Salary Components")
plt.tight_layout()
plt.show()
```

##Analysis:
 The heatmap shows the strength and direction of relationships between basic salary, allowances, bonuses, deductions, and net salary.

##Conclusion:
 The correlation analysis helps identify which salary components are most strongly related to net salary and supports better salary analysis.

##Q8.What is the distribution of employee salaries?

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    salary["Net_Salary"].dropna(),
    bins=15,
    kde=True
)

plt.title("Distribution of Employee Salaries")
plt.xlabel("Net Salary")
plt.ylabel("Number of Salary Records")
plt.tight_layout()
plt.show()
```

##Analysis:
 The histogram shows how salary records are distributed across different salary ranges.

##Conclusion:
 The chart helps HR understand common salary ranges and identify unusually high or low salary records.

##Analysis:

##Conclusion:
